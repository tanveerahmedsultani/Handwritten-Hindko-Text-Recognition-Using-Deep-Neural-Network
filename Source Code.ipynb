```python
import tensorflow as tf
from keras.layers import Dense, Dropout, Flatten, Conv2D, MaxPool2D, BatchNormalization
from keras.models import Sequential
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the dataset
train = pd.read_csv('aaa.csv')
test = pd.read_csv('testfile.csv')
```

```python
# Separate the target column (first column) from the rest of the data
train_y = train.iloc[:, 0]  # Assuming the target column is at index 0
train_x = train.iloc[:, 1:]  # Assuming the independent variables start from index 1
```

```python
# Separate the target column (first column) from the rest of the data
test_y = test.iloc[:, 0]  # Assuming the target column is at index 0
test_x = test.iloc[:, 1:]  # Assuming the independent variables start from index 1
```

```python
train_x = train_x.values.reshape(-1, 50, 50, 1)
```

```python
test_x = test_x.values.reshape(-1, 50, 50, 1)
```

```python
model = Sequential()
model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(50, 50, 1)))
model.add(MaxPool2D((2, 2)))
model.add(Dropout(0.25))
model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(MaxPool2D((2, 2)))
model.add(Dropout(0.25))
model.add(Conv2D(128, (3, 3), activation='relu'))
model.add(MaxPool2D((2, 2)))
model.add(Dropout(0.25))
model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(21, activation='softmax'))  # Assuming 20 classes in the dataset
```

```python
# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Train the model
history = model.fit(train_x, train_y, epochs=50, batch_size=20, validation_data=(test_x, test_y))
```

```python
# Evaluate the model
_, train_accuracy = model.evaluate(train_x, train_y, verbose=0)
print("Training accuracy:", train_accuracy)
```

```python
_, validation_accuracy = model.evaluate(test_x, test_y, verbose=0)
print("Validation accuracy:", validation_accuracy)
```

```python
# Save the model
model.save("hcr.h5")
```

```python
# Make predictions on the test dataset
predictions = model.predict(test_x)
predicted_labels = np.argmax(predictions, axis=1)
```

```python
# Calculate accuracy per class
unique_classes = np.unique(train_y)
accuracy_per_class = {}
for class_label in unique_classes:
    class_indices = np.where(test_y == class_label)[0]
    class_predictions = predicted_labels[class_indices]
    class_accuracy = np.sum(class_predictions == class_label) / len(class_indices)
    accuracy_per_class[class_label] = class_accuracy
```

```python
# Print accuracy for each class
for class_label, accuracy in accuracy_per_class.items():
    print(f"Accuracy for class {class_label}: {accuracy}")
```
